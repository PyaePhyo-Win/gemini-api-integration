# Gemini Extended Thinking

This notebook demonstrates how to use the 'Extended Thinking' (reasoning) feature of Gemini models using the `google-genai` SDK.

```python
import os
import sys
from pathlib import Path
from dotenv import load_dotenv
from google import genai
from google.genai import types

# Add project root to sys.path to import from utils
project_root = next(
    (path for path in [Path.cwd(), *Path.cwd().parents] if (path / "requirements.txt").exists()),
    Path.cwd(),
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from utils.gemini_retry import generate_content_with_retry

load_dotenv()

# Initialize the Gemini Client
# Ensure GOOGLE_API_KEY is set in your .env file
client = genai.Client(api_key=os.getenv("GOOGLE_API_KEY"))

# Use a model that supports thinking
MODEL_ID = "gemini-3-flash-preview"
```

```python
def chat_with_thinking(prompt, include_thoughts=True):
    """Sends a prompt to Gemini with thinking enabled."""
    config = types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(include_thoughts=include_thoughts),
    )
    
    # Using the shared retry helper for robustness
    response = generate_content_with_retry(
        client=client,
        model=MODEL_ID,
        contents=prompt,
        config=config,
    )
    return response

def display_response(response):
    """Separates and displays thoughts and the final response."""
    for part in response.candidates[0].content.parts:
        if part.thought:
            print("\033[94m--- MODEL THOUGHTS ---\033[0m")
            print(part.text)
            print("\n" + "="*50 + "\n")
        elif part.text:
            print("\033[92m--- FINAL RESPONSE ---\033[0m")
            print(part.text)
```

## Example: Solving a Complex Logic Puzzle

Extended thinking is particularly useful for complex reasoning tasks, math problems, or coding challenges.

```python
prompt = """
If I have three apples and you take away two, how many apples do you have?
Explain your reasoning step-by-step.
"""

try:
    response = chat_with_thinking(prompt)
    display_response(response)
except Exception as e:
    print(f"Error: {e}")
    print("\nPlease ensure your GOOGLE_API_KEY is valid and has access to the experimental thinking model.")
```
